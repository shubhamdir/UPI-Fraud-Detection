```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


comparison_data = {
    "Model": ["Random Forest", "SVM", "XGBoost"],
    "Accuracy": [0.9934, 0.984900, 0.99455],
    "Precision": [0.6949, 0.345455, 0.869919],
    "Recall": [0.6150, 0.570000, 0.535000],
    "F1-Score": [0.6525, 0.430189, 0.662539],
    "ROC-AUC":[0.9310, 0.917799, 0.961377],
    "Average Precision": [0.6620, 0.389640, 0.688419],
}

model_comparison = pd.DataFrame(comparison_data)

# Display comparison table
print("MODEL PERFORMANCE COMPARISON")
print("=" * 75)

display(
    model_comparison.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}",
        "ROC-AUC": "{:.4f}",
        "Average Precision": "{:.4f}"
    })
)


# Identify the best-performing model for each metric
metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score",
    "ROC-AUC",
    "Average Precision"
]

print("\nBEST MODEL FOR EACH METRIC")
print("=" * 75)

for metric in metric_columns:
    best_index = model_comparison[metric].idxmax()
    best_model = model_comparison.loc[best_index, "Model"]
    best_score = model_comparison.loc[best_index, metric]

    print(f"{metric}: {best_model} ({best_score:.4f})")
```

# 17. Validate XGBoost Predictions Against the Classification Threshold

```python
# Generate predictions using the saved threshold
recalculated_predictions = (
    xgb_test_predictions["Fraud_Probability"]
    >= xgb_threshold_loaded
).astype(int)

# Compare saved predictions with recalculated predictions
prediction_matches = (
    xgb_test_predictions["Predicted_Fraud"]
    == recalculated_predictions
)

# Count matching and mismatching predictions
matching_count = prediction_matches.sum()
mismatching_count = (~prediction_matches).sum()

# Display validation results
print("XGBOOST PREDICTION VALIDATION")
print("=" * 45)

print("Classification Threshold:", xgb_threshold_loaded)
print("Total Predictions:", len(xgb_test_predictions))
print("Matching Predictions:", matching_count)
print("Mismatching Predictions:", mismatching_count)

print(
    "\nPrediction Consistency:",
    f"{(matching_count / len(xgb_test_predictions)) * 100:.2f}%"
)

if mismatching_count == 0:
    print("\nAll saved predictions match the classification threshold.")
else:
    print("\nSome predictions do not match the classification threshold.")
```

# 20. Compare False Positives and False Negatives

```python
import pandas as pd
from IPython.display import display as show_table

# Store error counts from confusion matrices
error_comparison = pd.DataFrame({
    "Model": ["Random Forest", "SVM", "XGBoost"],
    "False Positives": [54, 216, 16],
    "False Negatives": [77, 86, 93]
})

# Calculate total classification errors
error_comparison["Total Errors"] = (
    error_comparison["False Positives"]
    + error_comparison["False Negatives"]
)

# Sort models by total errors
error_comparison = error_comparison.sort_values(
    by="Total Errors"
).reset_index(drop=True)

# Display comparison table
show_table(error_comparison.style.hide(axis="index"))
```
